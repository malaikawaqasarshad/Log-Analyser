```python
with open("securitylogins.txt", "r") as file:
    for line in file:
        print(line.strip())

failed_logins = 0

with open("securitylogins.txt", "r") as file:
    for line in file:

        if "LOGIN_FAILED" in line:
            failed_logins += 1

print("Failed login attempts:", failed_logins)

with open("securitylogins.txt", "r") as file:
    for line in file:

        if "LOGIN_FAILED" in line:

            parts = line.split()

            print(parts)

failed_by_ip = {}

with open("securitylogins.txt", "r") as file:

    for line in file:

        if "LOGIN_FAILED" in line:

            parts = line.split()

            ip = parts[4].split("=")[1]

            if ip not in failed_by_ip:
                failed_by_ip[ip] = 0

            failed_by_ip[ip] += 1


print("Failed logins by IP:")

for ip, count in failed_by_ip.items():
    print(ip, ":", count)

print("\nSuspicious IP addresses:")

for ip, count in failed_by_ip.items():

    if count >= 3:
        print("⚠", ip, "-", count, "failed attempts")
```
